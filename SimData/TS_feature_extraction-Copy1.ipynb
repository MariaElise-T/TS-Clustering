```python
br_poor = pd.read_csv("bank_reserves_outputs_poor.csv", header=None)
br_middle = pd.read_csv("bank_reserves_outputs_middle.csv", header=None)
br_rich = pd.read_csv("bank_reserves_outputs_rich.csv", header=None)
```

```python
def import_ff_data(filename):
    expected_columns=155
    data = []
    with open(filename, 'r') as file:
        for line in file:
            row = line.strip().split(',')
            if len(row) < expected_columns:
                row += [np.nan] * (expected_columns - len(row))
            data.append(row)
    df = pd.DataFrame(data)
    def fill_last_valid(row):
        for i in range(1, len(row)):
            if pd.isna(row[i]):
                row[i] = row[i - 1]  
        return row
    df_filled = df.apply(fill_last_valid, axis=1)
    return df_filled
```

```python
ff_onfire = import_ff_data("forest_fire_outputs_onfire.csv")
print("check 1")
ff_fine = import_ff_data("forest_fire_outputs_fine.csv")
print("check 2")
ff_burned = import_ff_data("forest_fire_outputs_burned.csv")
```

# Principal Component Analysis

```python
def pca_var_reduction(ts0, ts1, ts2):
    sc = StandardScaler()
    pca = PCA(n_components=3)
    PCA_std_list = list()
    for i in np.arange(0, ts0.shape[0]):
    #for i in np.arange(0, 1):
        if(i % 2000==0):
            print("On iteration ", i, "at ", datetime.datetime.now())
        sample = pd.concat((ts0.iloc[i], ts1.iloc[i]), axis=1)
        sample = pd.concat((sample, ts2.iloc[i]), axis=1)
        sample = sample.T
        sample_std = sc.fit_transform(sample)
        sample_PCA_std = pca.fit_transform(sample_std)
        explained_var = pca.explained_variance_ratio_
        PCA_std_list.append(pd.Series(explained_var)) 
    PCA_std_df = pd.DataFrame(PCA_std_list)
    return PCA_std_df
```

```python
br_pca = pca_var_reduction(br_poor, br_middle, br_rich)
```

```python
ecv_pca = pca_var_reduction(ecv_active, ecv_jailed, ecv_quiet)
```
