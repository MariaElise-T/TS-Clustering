```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
raw_sil_scores = pd.read_csv('sil_scores_labels/raw_sil_scores.csv')
raw_sil_scores = raw_sil_scores[['abm', 'k', 'score']]
raw_sil_scores['method'] = 'no embed - kmeans'
```

```python
pca_sil_scores = pd.read_csv('sil_scores_labels/pca_sil_scores.csv')
pca_sil_scores = pca_sil_scores[['abm', 'k', 'score']]
pca_sil_scores['method'] = 'pca - kmeans'
```

```python
dae_sil_scores = pd.read_csv('sil_scores_labels/dae_sil_scores.csv')
dae_sil_scores = dae_sil_scores[['abm', 'k', 'score']]
dae_sil_scores['method'] = 'dae - kmeans'
```

```python
dcae_sil_scores = pd.read_csv('sil_scores_labels/dcae_sil_scores.csv')
dcae_sil_scores = dcae_sil_scores[['abm', 'k', 'score']]
dcae_sil_scores['method'] = 'dcae - kmeans'
```

```python
daep_sil_scores = pd.read_csv('sil_scores_labels/daep_sil_scores.csv')
daep_sil_scores = daep_sil_scores[['abm', 'k', 'score']]
daep_sil_scores['method'] = 'daep - kmeans'
```

```python
daep_sil_scores_spec = pd.read_csv('sil_scores_labels/daep_sil_scores_spectral.csv')
daep_sil_scores_spec = daep_sil_scores_spec[['abm', 'k', 'score']]
daep_sil_scores_spec['method'] = 'daep - spectral'
```

```python
daep_sil_scores_agglom_knn = pd.read_csv('sil_scores_labels/daep_sil_scores_agglom_knn.csv')
daep_sil_scores_agglom_knn = daep_sil_scores_agglom_knn[['abm', 'k', 'score']]
daep_sil_scores_agglom_knn['method'] = 'daep - agglom-knn'
```

```python
daep_sil_scores_agglom_cent = pd.read_csv('sil_scores_labels/daep_sil_scores_agglom_centroids.csv')
daep_sil_scores_agglom_cent = daep_sil_scores_agglom_cent[['abm', 'k', 'score']]
daep_sil_scores_agglom_cent['method'] = 'daep - agglom-centroids'
```

```python
df = pd.concat([raw_sil_scores, pca_sil_scores, dcae_sil_scores, dae_sil_scores, 
                daep_sil_scores, daep_sil_scores_spec, daep_sil_scores_agglom_knn, daep_sil_scores_agglom_cent])
```

```python
daep_sil_scores_spec
```

```python
subset = df[
    (df['k'] == 3) &
    (df['method'].isin([
        'no embed - kmeans', 'pca - kmeans', 'dae - kmeans', 'dcae - kmeans',
        'daep - kmeans'
    ]))
]
```

```python
subset
```

```python
plt.figure(figsize=(10, 6))
sns.barplot(data=subset, x="abm", y="score", hue="method")

# Customize the plot
plt.xlabel("Model")
plt.ylabel("Silhouette Score")
plt.title("K-Means Clustering; k=3")
plt.legend(title="Feature Embedding")
plt.ylim(0, 1)
plt.savefig("kmeans_embedding_k3_comp.png", format="png")
plt.show()
```

```python
subset = df[
    (df['abm'] == 'BR') &
    (df['method'].isin([
        'no embed - kmeans', 'pca - kmeans', 'dae - kmeans', 'dcae - kmeans',
        'daep - kmeans'
    ]))
]
```

```python
subset
```

```python
plt.figure(figsize=(8,6))

for method, group in subset.groupby('method'):
    plt.plot(
        group['k'],
        group['score'],
        marker='o',
        linestyle='solid',
        label=method
    )

plt.xlabel("Number of Clusters")
plt.ylabel("Silhouette Score")
plt.title("Bank Reserves")
plt.legend(title='Method')
plt.tight_layout()
plt.savefig('br_size_of_k_silhouette_new.png')
plt.show()
```
