```python
import pandas as pd
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
from tslearn.utils import to_time_series_dataset
```

```python
raw_sil_scores = pd.read_csv('sil_scores_labels/raw_sil_scores.csv')
raw_sil_scores = raw_sil_scores[['abm', 'k', 'score']]
raw_sil_scores['method'] = 'no embed - kmeans'
```

```python
pca_sil_scores = pd.read_csv('sil_scores_labels/pca_sil_scores.csv')
pca_sil_scores = pca_sil_scores[['abm', 'k', 'score']]
pca_sil_scores['method'] = 'pca - kmeans'
```

```python
dae_sil_scores = pd.read_csv('sil_scores_labels/dae_sil_scores.csv')
dae_sil_scores = dae_sil_scores[['abm', 'k', 'score']]
dae_sil_scores['method'] = 'dae - kmeans'
```

```python
dcae_sil_scores = pd.read_csv('sil_scores_labels/dcae_sil_scores.csv')
dcae_sil_scores = dcae_sil_scores[['abm', 'k', 'score']]
dcae_sil_scores['method'] = 'dcae - kmeans'
```

```python
daep_sil_scores = pd.read_csv('sil_scores_labels/daep_sil_scores.csv')
daep_sil_scores = daep_sil_scores[['abm', 'k', 'score']]
daep_sil_scores['method'] = 'daep - kmeans'
```

```python
daep_sil_scores_spec = pd.read_csv('sil_scores_labels/daep_sil_scores_spectral.csv')
daep_sil_scores_spec = daep_sil_scores_spec[['abm', 'k', 'score']]
daep_sil_scores_spec['method'] = 'daep - spectral'
```

```python
daep_sil_scores_agglom_knn = pd.read_csv('sil_scores_labels/daep_sil_scores_agglom_knn.csv')
daep_sil_scores_agglom_knn = daep_sil_scores_agglom_knn[['abm', 'k', 'score']]
daep_sil_scores_agglom_knn['method'] = 'daep - agglom-knn'
```

```python
daep_sil_scores_agglom_cent = pd.read_csv('sil_scores_labels/daep_sil_scores_agglom_centroids.csv')
daep_sil_scores_agglom_cent = daep_sil_scores_agglom_cent[['abm', 'k', 'score']]
daep_sil_scores_agglom_cent['method'] = 'daep - agglom-centroids'
```

```python
df = pd.concat([raw_sil_scores, pca_sil_scores, dcae_sil_scores, dae_sil_scores, 
                daep_sil_scores, daep_sil_scores_spec, daep_sil_scores_agglom_knn, daep_sil_scores_agglom_cent])
```

```python
subset = df[
    (df['k'] == 3) &
    (df['method'].isin([
        'no embed - kmeans', 'pca - kmeans', 'dae - kmeans', 'dcae - kmeans',
        'daep - kmeans'
    ]))
]
```

```python
subset
```

```python
plt.figure(figsize=(10, 6))
sns.barplot(data=subset, x="abm", y="score", hue="method")

# Customize the plot
plt.xlabel("Model")
plt.ylabel("Silhouette Score")
plt.title("K-Means Clustering; k=3")
plt.legend(title="Feature Embedding")
plt.ylim(0, 1)
plt.savefig("kmeans_embedding_k3_comp.png", format="png")
plt.show()
```

```python
def ts_cluster_visualization2(y_pred, df, n_clusters, plot_title=None):
    """
    Plot only the cluster center (mean) curves for all clusters on a single plot.

    Parameters
    ----------
    y_pred : array-like of shape (n_samples,)
        Cluster labels for each time series (assumed 0..n_clusters-1).
    df : array-like of shape (n_samples, T)
        Time-series data; rows = series, columns = timesteps.
        Can be a NumPy array or a pandas DataFrame.
    n_clusters : int
        Number of clusters.
    plot_title : str, optional
        Title for the plot.
    """
    # Ensure we can index like a NumPy array
    X = np.asarray(df)
    ts_size = X.shape[1]
    ts_max = np.nanmax(X)

    plt.figure(figsize=(9, 5))
    for c in range(n_clusters):
        mask = (y_pred == c)
        if not np.any(mask):
            continue  # skip empty clusters gracefully
        center = np.nanmean(X[mask], axis=0)
        plt.plot(center.ravel(), linewidth=2, label=f"Cluster {c}")

    plt.xlim(0, ts_size - 1)
    plt.ylim(0, ts_max)
    if plot_title:
        plt.title(plot_title)
    plt.xlabel("Timestep")
    plt.ylabel("Value")
    plt.legend(frameon=False)
    plt.tight_layout()
    #plt.show()
```

```python
poor = pd.read_csv("SimData/bank_reserves_outputs_poor.csv", header=None)
X_train = to_time_series_dataset(poor)
```

```python
labels_df = pd.read_json("raw_labels.ndjson", lines=True)
method = "raw"      
abm = "BR"          
k = 7                #

row = labels_df[(labels_df["method"] == method) &
                (labels_df["abm"] == abm) &
                (labels_df["k"] == k)]

if not row.empty:
    labels = row.iloc[0]["labels"]
    print(f"Loaded {len(labels)} labels for {method.upper()} ({abm}), k={k}")
else:
    print("No matching entry found.")
```

```python
ts_cluster_visualization2(np.array(labels), X_train, 7, "Bank Reserves, No Embedding, K-Means")
plt.savefig("br_motifs_noembed_kmeans.png", format="png")
```
