```python
def ts_cluster_visualization2(y_pred, df, n_clusters, plot_title=None):
    """
    Plot only the cluster center (mean) curves for all clusters on a single plot.

    Parameters
    ----------
    y_pred : array-like of shape (n_samples,)
        Cluster labels for each time series (assumed 0..n_clusters-1).
    df : array-like of shape (n_samples, T)
        Time-series data; rows = series, columns = timesteps.
        Can be a NumPy array or a pandas DataFrame.
    n_clusters : int
        Number of clusters.
    plot_title : str, optional
        Title for the plot.
    """
    # Ensure we can index like a NumPy array
    X = np.asarray(df)
    ts_size = X.shape[1]
    ts_max = np.nanmax(X)

    plt.figure(figsize=(9, 5))
    for c in range(n_clusters):
        mask = (y_pred == c)
        if not np.any(mask):
            continue  # skip empty clusters gracefully
        center = np.nanmean(X[mask], axis=0)
        plt.plot(center.ravel(), linewidth=2, label=f"Cluster {c}")

    plt.xlim(0, ts_size - 1)
    plt.ylim(0, ts_max)
    if plot_title:
        plt.title(plot_title)
    plt.xlabel("Timestep")
    plt.ylabel("Value")
    plt.legend(frameon=False)
    plt.tight_layout()
    #plt.show()
```

```python
def import_ff_data(filename):
    expected_columns=155
    data = []
    with open(filename, 'r') as file:
        for line in file:
            row = line.strip().split(',')
            if len(row) < expected_columns:
                row += [np.nan] * (expected_columns - len(row))
            data.append(row)
    df = pd.DataFrame(data)
    def fill_last_valid(row):
        for i in range(1, len(row)):
            if pd.isna(row[i]):
                row[i] = row[i - 1]  
        return row
    df_filled = df.apply(fill_last_valid, axis=1)
    return df_filled
```

```python
onfire = import_ff_data("SimData/forest_fire_outputs_onfire.csv")
burned = import_ff_data("SimData/forest_fire_outputs_burned.csv")
fine = import_ff_data("SimData/forest_fire_outputs_fine.csv")
```

```python
def ts_cluster_visualization3(y_pred, df, n_clusters, plot_title=None, ylim=1):
    """
    Plot only the cluster center (mean) curves for all clusters on a single plot.

    Parameters
    ----------
    y_pred : array-like of shape (n_samples,)
        Cluster labels for each time series (assumed 0..n_clusters-1).
    df : array-like of shape (n_samples, T)
        Time-series data; rows = series, columns = timesteps.
        Can be a NumPy array or a pandas DataFrame.
    n_clusters : int
        Number of clusters.
    plot_title : str, optional
        Title for the plot.
    """
    # Ensure we can index like a NumPy array
    X = np.asarray(df)
    ts_size = X.shape[1]
    ts_max = np.nanmax(X)

    plt.figure(figsize=(9, 5))
    for c in range(n_clusters):
        mask = (y_pred == c)
        if not np.any(mask):
            continue  # skip empty clusters gracefully
        center = np.nanmean(X[mask], axis=0)
        plt.plot(center.ravel(), linewidth=2, label=f"Cluster {c}")

    plt.xlim(0, ts_size - 1)
    plt.ylim(0, ylim)
    if plot_title:
        plt.title(plot_title)
    plt.xlabel("Timestep")
    plt.ylabel("Value")
    plt.legend(frameon=False)
    plt.tight_layout()
    #plt.show()
```

```python
def to_numeric_df(df):
    return df.apply(pd.to_numeric, errors='coerce')

onfire = to_numeric_df(onfire)
burned = to_numeric_df(burned)
fine   = to_numeric_df(fine)

df_sum = onfire + burned + fine

sum_first_col = df_sum.iloc[:, 0]

onfire = onfire.div(sum_first_col, axis=0)
burned = burned.div(sum_first_col, axis=0)
fine   = fine.div(sum_first_col, axis=0)
```
